### Case - Cancelamento de Clientes

Você foi contratado por uma empresa com mais de 800 mil clientes para um projeto de Dados. Recentemente a empresa percebeu que da sua base total de clientes, a maioria são clientes inativos, ou seja, que já cancelaram o serviço.

Precisando melhorar seus resultados ela quer conseguir entender os principais motivos desses cancelamentos e quais as ações mais eficientes para reduzir esse número.

Base de dados e arquivos: https://drive.google.com/drive/folders/1uDesZePdkhiraJmiyeZ-w5tfc8XsNYFZ?usp=drive_link

```python
# Passo a passo do Projeto
# Passo 1: Importar a base de dados
import pandas
table = pandas.read_csv("cancelamentos.csv")

# Passo 2: Visualizar a base de dados
    # Entender as informações disponíveis
    # Possíveis problemas/erros na base de dados
table = table.drop(columns="CustomerID")
display(table)
```

```python
# Passo 3: Corrigir os problemas da base de dados (Tratamento de Dados)
display(table.info())
    # Valores em formatos errados -> Okay
    # Informações vazias -> 5 linhas com informações vazias
table = table.dropna()

display(table.info())

```

```python
# Passo 4: Análise inicial (Entender quantos clientes cancelaram)

# Quantidade
display(table["cancelou"].value_counts())

# Porcentagem
display(table["cancelou"].value_counts(normalize=True))
```

```python
# Passo 5: Análise detalhada (Causa dos cancelamentos dos clientes)
import plotly.express as px

# Criar o gráfico
for column in table.columns:
    graph = px.histogram(table,x=column, color="cancelou")

    # Exibir o gráfico
    graph.show()
```

```python
# Todo mundo do contrato mensal cancelou o serviço
    # Vamos dar desconto no contrato anual, trimestral

# Ligações call center acima de 4, todo mundo cancelou
    # tem algum problema que não estamos conseguindo resolver
    # se o cliente ligar 3x pro call center, acende um alerta vermelho

# Atraso de pagamento acima de 20 dias, cliente cancela
    # se o cliente atrasar mais de 15 dias no pagamento, alerta vermelho

# Cancelamento = 56%

# Filtrar a base de dados

#duracao_contrato -> diferente de Monthly
condition = table["duracao_contrato"] != "Monthly"
table = table[condition]

#ligacoes_callcenter -> menores ou iguais a 4
condition = table["ligacoes_callcenter"] <= 4
table = table[condition]

#dias_atraso -> menores ou iguais a 20
condition = table["dias_atraso"] <= 20
table = table[condition]

#porcentagem
display(table["cancelou"].value_counts(normalize=True))
```
